# Get Men's Barttorvik Ratings

Get the Barttorvik ratings from previous seasons for features about previous team strength. Assumes all data has been copied from website (https://barttorvik.com/team-tables_each.php) to Excel, as specified in the steps below. 

Steps before this file:

1. Copy from the website, only including games before tournament start date
2. Paste as text into Excel
3. Save into folder

```python
SEASON = 2025
INCLUDE_CURRENT_SEASON = False
```

```python
import pandas as pd

pd.set_option('display.max_columns', 100)

def load_data(season: int) -> pd.DataFrame:
    data = pd.read_excel(f'../data/unprocessed/mens_barttorvik/barttorvik_{season}.xlsx')
    data.columns = data.columns.str.upper()

    return data

df = pd.concat(
    [
        load_data(season)
        for season in range(2012, SEASON + INCLUDE_CURRENT_SEASON)
        if season != 2020  # cancelled
    ],
    ignore_index=True,
)

df.insert(0, 'Season', df.pop('YEAR'))

df
```

Create new columns

```python
df.insert(df.columns.get_loc('WINS'), 'WIN%', df['WINS']/(df['GAMES']))
df.insert(df.columns.get_loc('ADJ DE') + 1, 'ADJ EM', df['ADJ OE'] - df['ADJ DE'])

df
```

Remove some columns

```python
df = df.loc[
    : , 
    ~(
        (
            df.columns.isin([
                ' ', 
                'TEAM.1', 
                'RECORD', 
                'AVG HGT.', 
                'OP. FT%', 
                'PPP OFF.', 
                'PPP DEF.',
                '2P %',
                '2P % D',
                '3P %',
                '3P % D',
                'PAKE',  # DO NOT USE PAKE OR PASE; they are calculated post tourney despite being in earlier data
                'PASE',  # DO NOT USE PAKE OR PASE; they are calculated post tourney despite being in earlier data
                'WINS',
                'GAMES',
            ])
        ) | 
        (
            df.columns.str.contains('UNNAMED')
        )
    )
]

df
```
