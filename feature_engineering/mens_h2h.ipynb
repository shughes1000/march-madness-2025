# Get Men's Head-to-Head

Get head-to-head records and records against common opponents. 

### Head-to-Head

```python
SEASON = 2025
INCLUDE_CURRENT_SEASON = False
```

```python
import pandas as pd

pd.set_option('display.max_columns', 100)

df = pd.concat(
    [
        pd.read_parquet(fr'..\data\unprocessed\mens_sports_reference\full_season_sports_reference_{season}.parquet')
        .assign(Season=season)
        for season in range(2012, SEASON + INCLUDE_CURRENT_SEASON)
        if season != 2020  # tournament was cancelled
    ],
    ignore_index=True
)

df.insert(0, 'Season', df.pop('Season'))

df = df.loc[df['NCAA Tournament'] == 0, :].reset_index(drop=True)  # only include games before NCAA Tournament

df
```

```python
def rescale_weights(arr, minimum, maximum):
    """
    Rescale the weights array to match desired weights.
    Assumes the data is already transformed where a 1 point win is 1.00, a blowout is 1.50, and a tie is 0.50.
    Ties will be weighted as half the minimum.
    """

    arr_scaled = ((arr - 1.00) / 0.50) * (maximum - minimum) + minimum
    arr_scaled[arr == 0.50] = minimum / 2
    return arr_scaled

# make sure losses have negative adjusted score differential in this file
df['Adjusted Score Differential'] = rescale_weights(df['Adjusted Score Differential'], 0.50, 1.00)*df['Result']

df
```

```python
df_h2h = (
    df
    .groupby(['Season', 'Team', 'Opponent'])
    .agg(
        h2h_sum=('Adjusted Score Differential', 'sum'),
        h2h_count=('Result', 'count')
    )
    .reset_index()
    .rename(columns={'Team': 'Team A', 'Opponent': 'Team B'})
)

# add 1/4 to denominator so that a team that beats opponent multiple times is better accounted for
df_h2h['Head to Head'] = df_h2h['h2h_sum'] / (df_h2h['h2h_count'] + 1/4)

df_h2h
```

### Common Opponents

```python
df_h2h['h2h_avg'] = df_h2h['h2h_sum']/df_h2h['h2h_count']

df_h2h
```

```python
def get_opponents_results(season, team):
    data = df_h2h.loc[
        (df_h2h['Season'] == season) & 
        (df_h2h['Team A'] == team), 
        :
    ]
    
    return dict(zip(
        data['Team B'],
        data['h2h_avg']
    ))
```

```python
from tqdm.autonotebook import tqdm

season_team_opponents = {
    season: {
        team: get_opponents_results(season, team) 
        for team in df_h2h.loc[df_h2h['Season'] == season, 'Team A'].unique()
    } 
    for season in tqdm(df_h2h['Season'].unique())
}

len(season_team_opponents)
```

```python
import numpy as np

def get_difference(season, team_a, team_b):
    team_a_dict = season_team_opponents[season][team_a]
    team_b_dict = season_team_opponents[season][team_b]

    common_keys = team_a_dict.keys() & team_b_dict.keys()

    if len(common_keys) == 0:
        return np.nan, np.nan
    
    commons_opps_results = tuple(team_a_dict[key] - team_b_dict[key] for key in common_keys)

    return np.sum(commons_opps_results), (len(commons_opps_results))
```

```python
common_opps = [
    (season, team_a, team_b, *get_difference(season, team_a, team_b))
    for season in tqdm(df_h2h['Season'].unique())
    for team_a in df_h2h.loc[df_h2h['Season'] == season, 'Team A'].unique()
    for team_b in df_h2h.loc[df_h2h['Season'] == season, 'Team A'].unique()  # could use Team B but shouldn't matter
    if team_a != team_b
]

common_opps[:5]
```

```python
df_common_opps = pd.DataFrame(
    common_opps,
    columns=['Season', 'Team A', 'Team B', 'common_opps_sum', 'common_opps_count'],
)

df_common_opps
```

```python
df_h2h = pd.merge(
    df_h2h.drop(columns=['h2h_avg']),
    df_common_opps,
    how='outer',
    on=['Season', 'Team A', 'Team B'],
)

df_h2h
```

```python
# add 1/4 to denominator so that a team that beats opponent multiple times is better accounted for
df_h2h['Common Opps'] = df_h2h['common_opps_sum']/(df_h2h['common_opps_count'] + 1/4)

df_h2h
```

```python
df_h2h.loc[(df_h2h['Season'] == 2024) & df_h2h['Team A'].str.contains('Connecticut'), :]
```

Remove all the rows without data to reduce the dataframe size

```python
df_h2h = df_h2h.loc[(df_h2h['Head to Head'].notna()) | (df_h2h['Common Opps'].notna()), :].reset_index(drop=True)

df_h2h
```

```python
df_h2h.drop(columns=['h2h_sum', 'h2h_count', 'common_opps_sum', 'common_opps_count'], inplace=True)

df_h2h
```

```python
df_h2h.to_parquet('../data/preprocessed/mens_h2h/h2h.parquet')

'Done'
```
